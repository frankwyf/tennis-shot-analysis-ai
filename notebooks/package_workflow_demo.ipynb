````python
from tennis_shot_analysis import AnalysisConfig, run_analysis
from tennis_shot_analysis.visualize import plot_analysis
````

````python
config = AnalysisConfig(
    use_mediapipe_video=False,
    frame_count=80,
    duration_s=1.6,
)

landmarks, summary, label = run_analysis(config)
summary.to_series()
````

````python
plot_analysis(landmarks, summary, data_source_label=label, show=True)
````
